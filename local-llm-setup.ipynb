```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. LM Studio 로컬 서버 주소를 base_url로 지정하여 모델 선언
llm = ChatOpenAI(
    base_url="http://localhost:1234/v1",
    api_key="lm-studio",
    temperature=0.7
)

# 2. 프롬프트 템플릿 정의
prompt = ChatPromptTemplate.from_template("{topic}에 대해 2문장으로 요약해줘.")

# 3. LCEL 기반 파이프라인 생성 (| 연산자 사용)
chain = prompt | llm | StrOutputParser()

# 4. 체인 호출 및 실행
response = chain.invoke({"topic": "너가 누구인지"})
print(response)
```
